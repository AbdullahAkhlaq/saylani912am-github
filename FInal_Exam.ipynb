# Question 01: File Analysis and Word Search 

```python
def analyze_file(file_name):
    try:
        with open(file_name, 'r', encoding='utf-8') as file:
            content = file.read() 
            num_characters = len(content)
            num_words = len(content.split())  
            num_lines = len(content.splitlines())  
            
            return {
                'characters': num_characters,
                'words': num_words,
                'lines': num_lines
            }
    except FileNotFoundError:
        print(f"The file '{file_name}' was not found.")
        return None
    except Exception as e:
        print(f"Something is going wrong: {e}")
        return None


def search_word(file_name, word):

    try:
        with open(file_name, 'r', encoding='utf-8') as file:
            content = file.read()
            word_count = content.lower().split().count(word.lower()) 
            return word_count
    except FileNotFoundError:
        print(f"The file '{file_name}' was not found.")
        return -1
    except Exception as e:
        print(f"Something is going wrong: {e}")
        return -1


# Example usage
if __name__ == "__main__":
    analysis = analyze_file('Abdullah.txt')
    if analysis is not None:
        print("File Analysis:", analysis)

    count = search_word('Abdullah.txt', 'Abdullah')
    if count != -1:
        print(f"The word 'Abdullah' appears {count} times.")

```

# Question 2: Library Management System using Python OOP 

```python
class Book:
    
    def __init__(self, title, author, isbn):
        self.title = title
        self.author = author
        self.isbn = isbn
        self.is_borrowed = False  

    def borrow(self):
        if not self.is_borrowed:
            self.is_borrowed = True
        else:
            raise ValueError("This book is  borrowed.")

    def return_book(self):
        if self.is_borrowed:
            self.is_borrowed = False
        else:
            raise ValueError("This book is not borrowed.")


class DigitalBook(Book):
    
    def __init__(self, title, author, isbn, file_format):
        super().__init__(title, author, isbn)
        self.file_format = file_format  

    def borrow(self):
        super().borrow()
        print("You can read this book online")


class AudioBook(Book):
    
    def __init__(self, title, author, isbn, duration):
        super().__init__(title, author, isbn)
        self.duration = duration  

    def borrow(self):
        super().borrow()
        print("Audiobook is available")


class User:
    
    def __init__(self, user_id, name):
        self.user_id = user_id
        self.name = name
        self._borrowed_books = []  

    def borrow_book(self, book):
        if book.is_borrowed:
            raise ValueError("this book is  borrowed.")
        book.borrow()
        self._borrowed_books.append(book)

    def return_book(self, book):
        if book in self._borrowed_books:
            book.return_book()
            self._borrowed_books.remove(book)
        else:
            raise ValueError("You haven't borrowed this book.")


class Library:
    
    def __init__(self, name):
        self.name = name
        self.books = []  
        self.users = []  

    def add_book(self, book):
      
        self.books.append(book)

    def register_user(self, user):
       
        self.users.append(user)

    def lend_book(self, user_id, isbn):
      
        user = next((u for u in self.users if u.user_id == user_id), None)
        book = next((b for b in self.books if b.isbn == isbn), None)
        
        if user is None:
            raise ValueError("User not found.")
        if book is None:
            raise ValueError("Book not found.")
        user.borrow_book(book)

    def receive_return(self, user_id, isbn):
     
        user = next((u for u in self.users if u.user_id == user_id), None)
        book = next((b for b in self.books if b.isbn == isbn), None)

        if user is None:
            raise ValueError("User not found.")
        if book is None:
            raise ValueError("Book not found.")
        user.return_book(book)


# Example usage
if __name__ == "__main__":
   
    library = Library("DHA Library")

  
    book1 = Book("2002", "Abdllah Akhlaq", "0321")
    book2 = DigitalBook("Cricket", "Randhawa", "456", "PDF")
    book3 = AudioBook("Pakistan", "Ahmed", "789", 8.5)

    library.add_book(book1)
    library.add_book(book2)
    library.add_book(book3)

    user1 = User("u001", "Abdullah")
    user2 = User("u002", "Ahmed")

    library.register_user(user1)
    library.register_user(user2)

    try:
        library.lend_book("u001", "0321") 
        library.lend_book("u001", "456")  
        library.lend_book("u002", "789") 

        library.lend_book("u002", "456")  

    except ValueError as e:
        print(e)

    try:
        library.receive_return("u001", "0321")  
        library.receive_return("u002", "456")  
    except ValueError as e:
        print(e)

    try:
        library.receive_return("u001", "0321")  
    except ValueError as e:
        print(e)
```
